````python
# Datatype Fraction

class fraction:

  def __init__(self, n, d):
    self.num = n
    self.den = d

  def __str__(self):  # Print automatically calls
    return '{}/{}'.format(self.num, self.den)

  def __add__(self, other): # Print automatically add
    temp_num = self.num * other.den + other.num * self.den
    temp_den = self.den * other.den
    return '{}/{}'.format(temp_num, temp_den)

  def __sub__(self, other): # Print automatically sub
    temp_num = self.num * other.den - other.num * self.den
    temp_den = self.den * other.den
    return '{}/{}'.format(temp_num, temp_den)

  def __mul__(self, other): # Print automatically multiply
    temp_num = self.num * other.num
    temp_den = self.den * other.den
    return '{}/{}'.format(temp_num, temp_den)

  def __truediv__(self, other): # Print automatically div
    temp_num = self.num * other.den
    temp_den = self.den * other.num
    return '{}/{}'.format(temp_num, temp_den)

  def convert_to_decimal(self):
    return self.num/self.den




x = fraction(3,4)
print(x)
y = fraction(5,6)
print(y)

print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x.convert_to_decimal())

````
